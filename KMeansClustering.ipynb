```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
```

```python
#Cleaning data 
data = pd.read_csv("FootballPlayerStats.csv", sep=";", encoding="ISO-8859-1")
data = data[["Player", "Pos", "Comp", "Age", "MP", "Starts", "Min", "Shots", "SoT", "SoT%", "ShoDist", "ShoPK", "DriSucc", "DriAtt", "DriSucc%", "DriPast", "Goals"]]

data = data.set_index("Player")

data["Age"] = data["Age"].fillna(data["Age"].mean())

for i in range(len(data)):
    if data.iloc[i, 1] == "Premier League":
        data.iloc[i, 1] = 2.7
    elif data.iloc[i, 1] == "La Liga":
        data.iloc[i, 1] = 2.52
    elif data.iloc[i, 1] == "Bundesliga":
        data.iloc[i, 1] = 3.19
    elif data.iloc[i, 1] == "Serie A":
        data.iloc[i, 1] = 2.92
    elif data.iloc[i, 1] == "Ligue 1":
        data.iloc[i, 1] = 2.52

#Using K Means Clustering
positions = data["Pos"]
data = data.drop("Pos", axis=1)

scaler = MinMaxScaler()
scaler.fit(data)

data = pd.DataFrame(scaler.transform(data), columns=data.columns)

model = KMeans(n_clusters=2, random_state=42)
model.fit(data)

print(model.predict(data[100:110]))
print(positions[100:110].tolist())
```
