```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
```

```python
#Cleaning data 
data = pd.read_csv("2021-2022FootballPlayerStats.csv", sep=";", encoding="ISO-8859-1")

data = data.set_index("Player")

data["Age"] = data["Age"].fillna(data["Age"].mean())

for i in range(len(data)):
    if data["Pos"][i] == "GK" or data["Pos"][i] == "GKMF":
        data["Pos"][i] = 0 #GK
    elif data["Pos"][i] == "DF":
        data["Pos"][i] = 1 #DF
    elif data["Pos"][i] == "DFMF" or data["Pos"][i] == "DFFW" or data["Pos"][i] =="MFDF":
        data["Pos"][i] = 2 #LT
    elif data["Pos"][i] == "MF":
        data["Pos"][i] = 3 #MF
    elif data["Pos"][i] == "MFFW" or data["Pos"][i] == "FWMF" or data["Pos"][i] == "FWDF":
        data["Pos"][i] = 4 #WG
    elif data["Pos"][i] == "FW":
        data["Pos"][i] = 5 #FW

columns_to_drop = ["Born", "Rk", "Squad", "Nation", "Comp"]

data = data.drop(columns=columns_to_drop)
```
