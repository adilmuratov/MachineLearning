```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
```

```python
#Cleaning data 
data = pd.read_csv("FootballPlayerStats.csv", sep=";", encoding="ISO-8859-1")
data = data[["Player", "Pos", "Comp", "Age", "MP", "Starts", "Min", "Shots", "SoT", "SoT%", "ShoDist", "ShoPK", "DriSucc", "DriAtt", "DriSucc%", "DriPast", "Goals"]]

data = data.set_index("Player")

data["Age"] = data["Age"].fillna(data["Age"].mean())

for i in range(len(data)):
    if data.iloc[i, 0] == "GK":
        data.iloc[i, 0] = 0.01
    elif data.iloc[i, 0] == "GKMF":
        data.iloc[i, 0] = 0.01
    elif data.iloc[i, 0] == "DF":
        data.iloc[i, 0] = 1
    elif data.iloc[i, 0] == "DFMF":
        data.iloc[i, 0] = 1.25
    elif data.iloc[i, 0] == "MFDF":
        data.iloc[i, 0] = 1.25
    elif data.iloc[i, 0] == "FWDF":
        data.iloc[i, 0] = 1.5
    elif data.iloc[i, 0] == "DFFW":
        data.iloc[i, 0] = 1.5
    elif data.iloc[i, 0] == "MF":
        data.iloc[i, 0] = 1.5
    elif data.iloc[i, 0] == "MFFW":
        data.iloc[i, 0] = 2
    elif data.iloc[i, 0] == "FWMF":
        data.iloc[i, 0] = 2
    elif data.iloc[i, 0] == "FW":
        data.iloc[i, 0] = 3

for i in range(len(data)):
    if data.iloc[i, 1] == "Premier League":
        data.iloc[i, 1] = 2.7
    elif data.iloc[i, 1] == "La Liga":
        data.iloc[i, 1] = 2.52
    elif data.iloc[i, 1] == "Bundesliga":
        data.iloc[i, 1] = 3.19
    elif data.iloc[i, 1] == "Serie A":
        data.iloc[i, 1] = 2.92
    elif data.iloc[i, 1] == "Ligue 1":
        data.iloc[i, 1] = 2.52

#Using Linear Regression
name_player = "Kylian Mbappé"

player_with_goals = data.loc[[name_player]]

goals = data["Goals"]
data = data.drop("Goals", axis=1)

index_player = data.index.get_loc(name_player)

scaler = MinMaxScaler()
scaler.fit(data)

data = pd.DataFrame(scaler.transform(data), columns=data.columns)

model = LinearRegression()
model.fit(data, goals)

player_without_goals = data.iloc[[index_player]]

print(player_with_goals)
predicted_goals = model.predict(player_without_goals)
print(name_player + ": " + str(predicted_goals[0]))
```
